## Library

```python
import yfinance as yf
from ta.volatility import AverageTrueRange
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Imputasi data yahoo finance

#### Imputasi Data

```python
ptro = yf.download("PTRO.JK", start="2015-01-01", end="2026-05-25")
ihsg = yf.download("^JKSE", start="2015-01-01", end="2026-01-25")
usd = yf.download("IDR=X", start="2015-01-01", end="2026-01-25")
oil = yf.download("CL=F", start="2015-01-01", end="2026-01-25")
```

## Pra-Processing Data 

#### Pembentukan Variabel Y

Y1 merupakan volatilitas window 7

Y2 merupakan volatilitas window 14

```python
log_return = np.log(
    ptro['Close'] / ptro['Close'].shift(1)
)

vol_7 = log_return.rolling(window=21).std()

vol_14 = log_return.rolling(window=14).std()

vol_7 = vol_7.dropna()
vol_14 = vol_14.dropna()

scaler_Y1 = MinMaxScaler()
scaler_Y2 = MinMaxScaler()

Y1 = pd.DataFrame(
    scaler_Y1.fit_transform(
        vol_7.values.reshape(-1,1)
    ),
    index=vol_7.index,
    columns=['Y1']
)

Y2 = pd.DataFrame(
    scaler_Y2.fit_transform(
        vol_14.values.reshape(-1,1)
    ),
    index=vol_14.index,
    columns=['Y2']
)
```

#### Transformasi Variabel

Volume PTRO

```python
(ptro['Volume'] == 0).sum()
```

```python
log_volume = np.log(
    ptro['Volume'] + 1
)

log_volume = log_volume.dropna()

scaler_volume = MinMaxScaler()

X_volume = pd.DataFrame(
    scaler_volume.fit_transform(
        log_volume.values.reshape(-1,1)
    ),
    index=log_volume.index,
    columns=['X_volume']
)
```

IHSG

```python
ihsg_return = np.log(
    ihsg['Close'] / ihsg['Close'].shift(1)
)

ihsg_return = ihsg_return.dropna()
scaler_ihsg = MinMaxScaler()

X_ihsg = pd.DataFrame(
    scaler_ihsg.fit_transform(
        ihsg_return.values.reshape(-1,1)
    ),
    index=ihsg_return.index,
    columns=['X_ihsg']
)
```

USD

```python
usd_return = np.log(
    usd['Close'] / usd['Close'].shift(1)
)

usd_return = usd_return.dropna()
scaler_usd  = MinMaxScaler()

X_usd = pd.DataFrame(
    scaler_usd.fit_transform(
        usd_return.values.reshape(-1,1)
    ),
    index=usd_return.index,
    columns=['X_usd']
)
```

OIL

```python
oil_return = np.log(
    oil['Close'] / oil['Close'].shift(1)
)

oil_return = oil_return.dropna()
scaler_oil  = MinMaxScaler()

X_oil = pd.DataFrame(
    scaler_oil.fit_transform(
        oil_return.values.reshape(-1,1)
    ),
    index=oil_return.index,
    columns=['X_oil']
)
```

#### Merge Data

```python
Data_7 = pd.concat(
    [
        Y1,
        X_volume,
        X_ihsg,
        X_usd,
        X_oil
    ],
    axis=1,
    join='inner'
)
```
